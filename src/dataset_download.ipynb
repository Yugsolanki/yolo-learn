```python
from roboflow import Roboflow
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
API_KEY = os.getenv("ROBOFLOW_API_KEY")
if not API_KEY:
    raise RuntimeError("Missing ROBOFLOW_API_KEY")

rf = Roboflow(api_key=API_KEY)

TARGETS = [
    {"workspace": "muhammadrizkiproject", "project": "clock-r8zgr"},
    {"workspace": "clock-tvzts", "project": "clock-buxuu"},
    {"workspace": "2-olter", "project": "office-wall-clock"}
]

DOWNLOAD_FORMAT = "yolo26"
DATASET_PATH = "../datasets"

for target in TARGETS:
    ws_name = target["workspace"]
    proj_name = target["project"]
    
    print(f"\n--- Fetching: {ws_name}/{proj_name} ---")
    
    try:
        project = rf.workspace(ws_name).project(proj_name)
        versions_metadata = project.versions()
        
        if isinstance(versions_metadata, list):
            version_numbers = [v.version for v in versions_metadata]
        else:
            version_numbers = [v.version for v in versions_metadata.values()]
            
        print(f"Found {len(version_numbers)} version(s): {version_numbers}")
        
        for version_num in version_numbers:
            print(f"Downloading version {version_num}...")
            version = project.version(version_num)
            dataset = version.download(DOWNLOAD_FORMAT, location=f"{DATASET_PATH}/{ws_name}/{proj_name}/v{version_num}")
            print(f"Successfully downloaded version {version_num} to {dataset.location}")
    except Exception as e:
        print(f"Error processing {ws_name}/{proj_name}: {e}")
        
print("\nAll downloads completed.")
```
